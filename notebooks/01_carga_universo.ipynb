# Carga del universo de empresas

Lee la tab `data` de `data/empresas_data_final.xlsx` (vía `universo.py`) y construye el universo de inversión:

- **Ticker** y datos descriptivos (empresa, sector, país)
- **Market Cap** (col. F) → para el prior de equilibrio de Black-Litterman (π = δΣw)
- **RANKINGFinal** (col. AT) → base de las views relativas
- **Peso en Cartera** (col. AU) → destino de los pesos optimizados

> Nota: las columnas AN y AT comparten el mismo encabezado (`RANKINGFinal`), por lo que la
> extracción es posicional y no por nombre (ver docstring de `universo.py`).

```python
from universo import cargar_universo

universo = cargar_universo()
print(f"Universo: {len(universo)} empresas | Ranking {universo['Ranking'].min()}–{universo['Ranking'].max()}")
print(f"Columnas: {list(universo.columns)}")
universo.head(10)
```

```python
print("Empresas por sector:")
print(universo["Sector"].value_counts())
print("\nPeso en Cartera ya asignado:", universo["Peso"].notna().sum())
print("Market caps nulos:", universo["MarketCapUSD"].isna().sum())
```
